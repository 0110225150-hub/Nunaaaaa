```python
# TODO E1 — HAPUS
# Daftar kolom yang memuat data sensitif/PII (harus dihapus)
kolom_hapus = ['nama_lengkap', 'nik_dummy', 'no_hp', 'email', 'alamat_jalan']

# Membuang kolom yang ditentukan
aman = df.drop(columns=kolom_hapus)

# Menghitung jumlah kolom
jumlah_dibuang = len(kolom_hapus)
jumlah_tersisa = aman.shape[1]

# Cetak hasil
print(f"Jumlah kolom yang dibuang : {jumlah_dibuang}")
print(f"Jumlah kolom yang tersisa : {jumlah_tersisa}")
```

```python
# TODO E2 — HASH
# Tulis fungsi samarkan(nilai, garam) yang mengembalikan 12 karakter pertama
# dari sha256(garam + nilai). Kembalikan np.nan bila nilainya kosong.
# Terapkan pada kolom yang Anda beri tindakan 'hash'.
# Expected: nilai asli dan hasil hash berbeda, dan nilai asli yang sama menghasilkan hash yang sama

GARAM = 'ganti-dengan-kalimat-rahasia-tim'
def samarkan(nilai, garam):
    # Cek jika nilai kosong/NaN
    if pd.isna(nilai):
        return np.nan

    # Gabungkan garam + nilai (konversi nilai ke string terlebih dahulu)
    teks = str(garam) + str(nilai)

    # Hitung hash SHA-256 dan ambil 12 karakter pertama
    hash_obj = hashlib.sha256(teks.encode('utf-8'))
    return hash_obj.hexdigest()[:12]

# Kolom pengenal unik yang disamarkan menggunakan hash
kolom_hash = ['id_peserta']

# Terapkan fungsi samarkan ke DataFrame 'aman'
for col in kolom_hash:
    aman[col] = aman[col].apply(lambda x: samarkan(x, GARAM))

# Menampilkan 5 baris pertama untuk memastikan pengubahan berhasil
print(aman[kolom_hash].head())
```

```python
# TODO E3 — buktikan hash bersifat konsisten dan tidak terbaca
# a. Hash dua kali nilai yang sama, tunjukkan hasilnya identik
# b. Hash dua nilai berbeda, tunjukkan hasilnya berbeda

# Contoh nilai uji coba
nilai1 = 'P001'
nilai2 = 'P001'  # Nilai yang sama dengan nilai1
nilai3 = 'P002'  # Nilai yang berbeda

# a. Hash dua kali nilai yang sama
hash_a1 = samarkan(nilai1, GARAM)
hash_a2 = samarkan(nilai2, GARAM)

print("--- Uji A: Hash Dua Nilai yang Sama ---")
print(f"Hash 1 ({nilai1}) : {hash_a1}")
print(f"Hash 2 ({nilai2}) : {hash_a2}")
print(f"Hasil sama?     : {hash_a1 == hash_a2}")

print("\n--- Uji B: Hash Dua Nilai yang Berbeda ---")
# b. Hash dua nilai berbeda
hash_b1 = samarkan(nilai1, GARAM)
hash_b2 = samarkan(nilai3, GARAM)

print(f"Hash 1 ({nilai1}) : {hash_b1}")
print(f"Hash 3 ({nilai3}) : {hash_b2}")
print(f"Hasil berbeda?  : {hash_b1 != hash_b2}")
```

**Jawaban Anda:** kalau `GARAM` ikut ter-*commit* ke GitHub, apa yang masih terlindungi dan apa yang tidak?

_kalau GARAM ikut tercommit, nilai hash tetap tdk bisa langsung dideskripsi kembali ke data asli, tapi akan rentan terhadap Brute Force karena peretas yg tau string GARAM bisa membuat daftar jebakan lalu di hash utk dicocokkan dgn data asli, sehingga fungsi salt nya pun hilang/sia-sia._

```python
# TODO E4 — GENERALISASI

ACUAN = pd.Timestamp('2025-09-01')

# a. Hitung usia peserta pada tanggal acuan 1 September 2025 dengan mengambil dari df asli jika sudah terhapus di aman
tanggal_lahir_sumber = df['tanggal_lahir']
tanggal_lahir_sumber = pd.to_datetime(tanggal_lahir_sumber)
usia = (ACUAN - tanggal_lahir_sumber).dt.days // 365.25

# b. Buat kolom 'kelompok_umur' dengan batas: <25, 25-34, 35-44, 45-54, 55+
bins = [-float('inf'), 24, 34, 44, 54, float('inf')]
labels = ['<25', '25-34', '35-44', '45-54', '55+']
aman['kelompok_umur'] = pd.cut(usia, bins=bins, labels=labels)

# c. Buang kolom tanggal_lahir dan kelurahan dari 'aman' secara aman (ignore errors jika sudah dibuang)
aman = aman.drop(columns=['tanggal_lahir', 'kelurahan'], errors='ignore')

# d. Rapikan kolom kecamatan yang sebagian tertulis KAPITAL
aman['kecamatan'] = aman['kecamatan'].str.title()

# Memastikan output sesuai ekspektasi
print("Kelompok umur terbanyak:\n", aman['kelompok_umur'].value_counts())
print("\nJumlah kecamatan unik:", aman['kecamatan'].nunique())
print("Daftar kecamatan:", aman['kecamatan'].unique())
```

## Bagian F — Memeriksa sisa risiko

Menghapus nama belum tentu cukup. Untuk setiap kombinasi quasi-identifier, hitung berapa orang yang punya kombinasi sama. Nilai terkecil disebut **k**. Target kelas ini: **k ≥ 5**.

```python
# TODO F1 — tulis fungsi cek_k(data, kolom)
# Kembalikan Series berisi: k (grup terkecil), grup berisi 1 orang, total grup
def cek_k(data, kolom):
    grup = data.groupby(kolom, observed=True).size()
    grup_aktif = grup[grup > 0]

    k = grup_aktif.min()
    grup_1_orang = (grup_aktif == 1).sum()
    total_grup = len(grup_aktif)

    return pd.Series({
        'k': k,
        'grup berisi 1 orang': grup_1_orang,
        'total grup': total_grup
    })
```

```python
# TODO F2 — ukur risiko SEBELUM penanganan
# Gunakan data asli df dengan kolom:
#   ['kelurahan', 'jenis_kelamin', 'pendidikan_terakhir', 'tanggal_lahir']
# Expected: k = 1 dan SELURUH 620 baris berada di grup berisi satu orang
kolom_qi_sebelum = ['kelurahan', 'jenis_kelamin', 'pendidikan_terakhir', 'tanggal_lahir']
hasil_f2 = cek_k(df, kolom_qi_sebelum)
print(hasil_f2)
```

**Jawaban Anda:** keempat kolom itu tidak memuat nama maupun nomor telepon. Jelaskan dalam dua kalimat mengapa hasilnya tetap berbahaya.

_Kombinasi data spesifik tersebut dapat mengarah pada satu individu tertentu jika dicocokkan dengan data publik lain. Nilai (k = 1) pada seluruh 620 baris membuktikan bahwa setiap orang dalam dataset tersebut sangat mudah diidentifikasi kembali._

```python
# TODO F3 — ukur risiko SESUDAH generalisasi
# Gunakan 'aman' dengan kolom:
#   ['kecamatan', 'jenis_kelamin', 'pendidikan_terakhir', 'kelompok_umur']
# Expected: jumlah grup berisi satu orang turun banyak, tetapi k masih 1
kolom_qi_sesudah = ['kecamatan', 'jenis_kelamin', 'pendidikan_terakhir', 'kelompok_umur']
hasil_f3 = cek_k(aman, kolom_qi_sesudah)
print(hasil_f3)
```

```python
# TODO F4 — TANTANGAN: capai k >= 5
# Generalisasikan lebih jauh sampai k >= 5, dengan MEMPERTAHANKAN sebanyak mungkin informasi.
# Anda boleh: memperlebar kelompok umur, menggabungkan kecamatan kecil menjadi 'Lainnya',
# atau mengeluarkan satu kolom dari daftar quasi-identifier.
# Catat setiap pilihan dan dampaknya terhadap k.
# Petunjuk: ada lebih dari satu jawaban benar. Yang dinilai adalah alasannya.
bins_baru = [-float('inf'), 34, float('inf')]
labels_baru = ['<35', '35+']
aman['kelompok_umur_luas'] = pd.cut(usia, bins=bins_baru, labels=labels_baru)

counts = aman['kecamatan'].value_counts()
kecamatan_kecil = counts[counts < 10].index
aman['kecamatan_grouped'] = aman['kecamatan'].replace(kecamatan_kecil, 'Lainnya')

kolom_qi_f4 = ['kecamatan_grouped', 'jenis_kelamin', 'kelompok_umur_luas']
hasil_f4 = cek_k(aman, kolom_qi_f4)
print(hasil_f4)

```

**Jawaban Anda:** tulis kombinasi yang Anda pilih, nilai k yang dicapai, dan informasi apa yang Anda korbankan untuk mencapainya.

_Kombinasi yg dipilih: menggabungkan `kecamatan_grouped`, `jenis_kelamin`, `kelompok_umur_luas`_
_Nilai k yg diciptakan: k >= 5_  
_Informasi yg dikorbankan: kolom `pendidikan_terakhir` dikeluarkan dari quasi-identifier, rentang usia diperlebar menjadi dua kelompok (<35 dan 35+)_

```python
# TODO F5 — sel kecil pada tabel agregat

# a. Buat tabel silang jumlah peserta menurut kecamatan x kategori_usaha
tabel_silang = pd.crosstab(aman['kecamatan'], aman['kategori_usaha'])

# b. Hitung berapa sel yang terisi tetapi berisi kurang dari 5 peserta (bernilai 1, 2, 3, atau 4)
sel_kecil = ((tabel_silang > 0) & (tabel_silang < 5)).sum().sum()

print("Tabel Silang:")
print(tabel_silang)
print("\nJumlah sel terisi yang kurang dari 5 peserta:", sel_kecil)
```
