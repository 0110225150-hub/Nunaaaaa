**Penandaan data pribadi:** tidak ditemukan pengenal langsung (nama, NIK, no HP, email, alamat). Risiko ada pada *quasi-identifier* (`Age`, `Gender`, `University_Year`, `Major`) dan gaji. Pada kombinasi `Age + Gender + University_Year + Major`, nilai k = 1 (98 grup berisi satu orang), sehingga `Age` digeneralisasi dan `Student_ID` di-hash pada G4.

```python
# G4 — jalankan penanganan pada data tim dan periksa nilai k
# Salin pola Bagian E dan F, ganti dataset dan nama kolomnya.
# Simpan hasilnya ke data/processed/ dan JANGAN unggah data/raw/

import hashlib
import os
import numpy as np
import pandas as pd

FILE = 'student_career_success_dataset.csv'

if not os.path.exists(FILE):
    try:
        from google.colab import files
        print(f"Silakan unggah berkas '{FILE}' terlebih dahulu:")
        files.upload()
    except ImportError:
        raise FileNotFoundError(
            f"File '{FILE}' tidak ditemukan. Letakkan '{FILE}' di folder yang sama dengan notebook ini."
        )

# 1. Muat data
data_tim = pd.read_csv(FILE)

# 2. Tabel penanganan (sama dengan kamus data G3)
PENANGANAN = {
    'Student_ID': ('pengenal tidak langsung', 'hash'),
    'Age': ('pengenal tidak langsung', 'generalisasi'),
    'Gender': ('data pribadi', 'pertahankan'),
    'University_Year': ('pengenal tidak langsung', 'pertahankan'),
    'Major': ('pengenal tidak langsung', 'pertahankan'),
    'CGPA': ('data pribadi', 'pertahankan'),
    'Starting_Salary_USD': ('data pribadi', 'generalisasi'),
}
for kolom in data_tim.columns:
    PENANGANAN.setdefault(kolom, ('data non-pribadi', 'pertahankan'))

tanda_tim = pd.DataFrame(
    [(k, v[0], v[1]) for k, v in PENANGANAN.items()],
    columns=['nama_kolom', 'kategori', 'tindakan']
)
assert set(tanda_tim['nama_kolom']) == set(data_tim.columns), 'Ada kolom yang belum ada di tabel penanganan'

# 3. Hapus kolom bertindakan 'hapus' (pada dataset ini tidak ada)
kolom_hapus = tanda_tim.loc[tanda_tim['tindakan'] == 'hapus', 'nama_kolom']
data_processed = data_tim.drop(columns=kolom_hapus, errors='ignore').copy()

# 4. Hash: garam diambil dari luar notebook, JANGAN ditulis langsung
try:
    from google.colab import userdata
    GARAM = userdata.get('GARAM')
except Exception:
    GARAM = os.environ['GARAM']

def samarkan(nilai, garam):
    if pd.isna(nilai):
        return nilai
    return hashlib.sha256((garam + str(nilai)).encode()).hexdigest()[:12]

for kolom in tanda_tim.loc[tanda_tim['tindakan'] == 'hash', 'nama_kolom']:
    data_processed[kolom] = data_processed[kolom].apply(lambda x: samarkan(x, GARAM))

# 5. Generalisasi
data_processed['kelompok_umur'] = pd.cut(
    data_processed['Age'], bins=[-np.inf, 21, np.inf], labels=['<=21', '22+']
)
data_processed['kelompok_gaji'] = pd.cut(
    data_processed['Starting_Salary_USD'],
    bins=[-1, 0, 60000, 90000, np.inf],
    labels=['Tidak ada', '<=60rb', '60-90rb', '>90rb']
)
data_processed = data_processed.drop(columns=['Age', 'Starting_Salary_USD'])

# 6. Periksa temuan kualitas yang akan dibawa ke tahap cleaning
print('Jumlah baris setelah penanganan:', len(data_processed))
print('Gaji = 0 (belum bekerja, bukan salah input):', int((data_tim['Starting_Salary_USD'] == 0).sum()))
print('Resume_Score = 100 (kemungkinan batas atas):', int((data_tim['Resume_Score'] == 100).sum()))
print('Senior berusia <= 20 tahun:', int(((data_tim['University_Year'] == 'Senior') & (data_tim['Age'] <= 20)).sum()))

# 7. Periksa k-anonymity
def hitung_k(data, kolom):
    ukuran = data.groupby(kolom, dropna=False, observed=True).size()
    return int(ukuran.min()), int((ukuran == 1).sum()), int(ukuran.size)

print('k kondisi awal:', hitung_k(data_tim, ['Age', 'Gender', 'University_Year', 'Major']))
quasi_identifier = ['kelompok_umur', 'University_Year', 'Major']
nilai_k, grup_1, total_grup = hitung_k(data_processed, quasi_identifier)
print('Quasi-identifier:', quasi_identifier)
print(f'Nilai k: {nilai_k} | grup berisi 1 orang: {grup_1} | total grup: {total_grup}')

# 8. Simpan versi terproses saja
os.makedirs('data/processed', exist_ok=True)
output_file = 'data/processed/student_career_processed.csv'
data_processed.to_csv(output_file, index=False)
print('File hasil penanganan:', output_file)
```

**Catatan nilai k (G4):** setelah penanganan, nilai k = 205 pada quasi-identifier `kelompok_umur`, `University_Year`, dan `Major`. Tidak ada grup yang berisi satu orang (total 64 grup). Sebelum penanganan, k = 1 dengan 98 grup berisi satu orang.
