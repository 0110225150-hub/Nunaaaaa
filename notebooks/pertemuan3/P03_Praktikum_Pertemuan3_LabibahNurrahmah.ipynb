```python
# TODO D1 – susun tabel penandaan Anda sendiri
import pandas as pd

# Dictionary penandaan kolom (nama_kolom: (kategori, tindakan))
PENANDAAN = {
    # Pengenal Langsung (Direct Identifier)
    'nama_lengkap': ('pengenal langsung', 'hapus'),
    'nik': ('pengenal langsung', 'hapus'),
    'nomor_hp': ('pengenal langsung', 'hapus'),
    'email': ('pengenal langsung', 'hapus'),
    'nama_usaha': ('pengenal langsung', 'hapus'),

    # Data Spesifik (Sensitive Personal Data)
    'pendapatan_pribadi': ('data spesifik', 'generalisasi'),
    'skor_kredit': ('data spesifik', 'generalisasi'),
    'jumlah_anak': ('data spesifik', 'pertahankan'),
    'riwayat_kesehatan': ('data spesifik', 'hapus'),

    # Quasi-identifier
    'tanggal_lahir': ('quasi-identifier', 'generalisasi'),
    'kecamatan': ('quasi-identifier', 'pertahankan'),
    'pendidikan_terakhir': ('quasi-identifier', 'pertahankan'),

    # Pengenal Unik yang Dipseudonisimasi
    'id_peserta': ('pengenal langsung', 'hash')
}

# Membangun list data untuk DataFrame
data_tanda = []
for col in df.columns:
    if col in PENANDAAN:
        kategori, tindakan = PENANDAAN[col]
    else:
        kategori, tindakan = 'bukan data pribadi', 'pertahankan'

    data_tanda.append({
        'nama_kolom': col,
        'kategori': kategori,
        'tindakan': tindakan
    })

# Membuat DataFrame 'tanda'
tanda = pd.DataFrame(data_tanda)

# Mencetak jumlah kolom per kategori
print("--- Jumlah Kolom Per Kategori ---")
print(tanda['kategori'].value_counts())
```

## Bagian E — Menerapkan tiga tindakan

```python
# TODO E1 – HAPUS
# Ambil daftar kolom yang memiliki tindakan 'hapus' dari DataFrame 'tanda'
kolom_dibuang = tanda[tanda['tindakan'] == 'hapus']['nama_kolom'].tolist()

# Buat DataFrame baru 'aman' dengan menghapus kolom-kolom tersebut
aman = df.drop(columns=kolom_dibuang).copy()

# Cetak berapa kolom yang dibuang dan berapa yang tersisa
print(f"Jumlah kolom yang dibuang : {len(kolom_dibuang)}")
print(f"Jumlah kolom yang tersisa : {aman.shape[1]}")
print(f"Kolom yang dibuang        : {kolom_dibuang}")
```

```python
# TODO E2 – HASH
import hashlib
import numpy as np
import pandas as pd

GARAM = 'ganti-dengan-kalimat-rahasia-tim'

def samarkan(nilai, garam=GARAM):
    # Kembalikan np.nan jika nilai kosong (null / NaN)
    if pd.isna(nilai):
        return np.nan

    # Gabungkan garam + nilai (dikonversi ke string terlebih dahulu)
    gabungan = str(garam) + str(nilai)

    # Hitung SHA-256 hash dan ambil 12 karakter pertama
    hash_object = hashlib.sha256(gabungan.encode('utf-8'))
    return hash_object.hexdigest()[:12]

# Ambil daftar kolom yang memiliki tindakan 'hash'
kolom_hash = tanda[tanda['tindakan'] == 'hash']['nama_kolom'].tolist()

# Terapkan fungsi samarkan pada kolom-kolom tersebut di DataFrame 'aman'
for col in kolom_hash:
    if col in aman.columns:
        aman[col] = aman[col].apply(lambda x: samarkan(x, GARAM))

print("Proses hashing selesai untuk kolom:", kolom_hash)
```

```python
# TODO E3 – buktikan hash bersifat konsisten dan tidak terbaca

# Nilai sampel untuk pengujian
nilai_a1 = "P001"
nilai_a2 = "P001"
nilai_b  = "P002"

# a. Hash dua kali nilai yang sama, tunjukkan hasilnya identik
hash_a1 = samarkan(nilai_a1, GARAM)
hash_a2 = samarkan(nilai_a2, GARAM)

print("--- Test A: Nilai Sama ---")
print(f"Hash 1 ('{nilai_a1}') : {hash_a1}")
print(f"Hash 2 ('{nilai_a2}') : {hash_a2}")
print(f"Apakah hasil identik? : {hash_a1 == hash_a2}")

# b. Hash dua nilai berbeda, tunjukkan hasilnya berbeda
hash_b = samarkan(nilai_b, GARAM)

print("\n--- Test B: Nilai Berbeda ---")
print(f"Hash ('{nilai_a1}') : {hash_a1}")
print(f"Hash ('{nilai_b}')  : {hash_b}")
print(f"Apakah hasil berbeda? : {hash_a1 != hash_b}")
```

**Jawaban Anda:** kalau `GARAM` ikut ter-*commit* ke GitHub, apa yang masih terlindungi dan apa yang tidak?

- Yang masih terlindungi:
  Penyerang tidak bisa langsung mengembalikan (reverse) atau melakukan dekripsi hash secara langsung untuk membaca teks asli, karena fungsi SHA-256 bersifat satu arah (one-way function)

- Yang tidak terlindungi:
  Penyerang yang mengetahui nilai GARAM dapat menggunakan dictionary attack atau rainbow table dengan cara melakukan pencocokan hash dari daftar kemungkinan nilai input (misalnya mencoba semua pola NIK, ID peserta, atau nomor telepon) hingga menghasilkan string hash yang sama persis.

```python
# TODO E4 – GENERALISASI
import pandas as pd

ACUAN = pd.Timestamp('2025-09-01')

# a. Hitung usia peserta pada tanggal acuan 1 September 2025
usia = (ACUAN - aman['tanggal_lahir']).dt.days // 365

# b. Buat kolom 'kelompok_umur' dengan batas: <25, 25-34, 35-44, 45-54, 55+
bins = [-1, 24, 34, 44, 54, 150]
labels = ['<25', '25-34', '35-44', '45-54', '55+']
aman['kelompok_umur'] = pd.cut(usia, bins=bins, labels=labels)

# c. Buang kolom tanggal_lahir dan kelurahan dari 'aman'
kolom_dibuang = [col for col in ['tanggal_lahir', 'kelurahan'] if col in aman.columns]
aman = aman.drop(columns=kolom_dibuang)

# d. Rapikan kolom kecamatan yang sebagian tertulis KAPITAL
aman['kecamatan'] = aman['kecamatan'].str.title()

# Tampilkan informasi verifikasi
print("--- Frekuensi Kelompok Umur ---")
print(aman['kelompok_umur'].value_counts())
print(f"\nJumlah kecamatan unik: {aman['kecamatan'].nunique()}")
```

## Bagian F — Memeriksa sisa risiko

Menghapus nama belum tentu cukup. Untuk setiap kombinasi quasi-identifier, hitung berapa orang yang punya kombinasi sama. Nilai terkecil disebut **k**. Target kelas ini: **k ≥ 5**.

```python
# TODO F1 – tulis fungsi cek_k(data, kolom)
import pandas as pd

def cek_k(data, kolom):
    # Hitung ukuran/anggota untuk setiap kombinasi grup quasi-identifier
    grup = data.groupby(kolom).size()

    k_min = grup.min()                         # Nilai k (grup terkecil)
    grup_1_orang = (grup == 1).sum()            # Jumlah grup yang hanya berisi 1 orang
    total_grup = len(grup)                      # Total kombinasi grup unik

    return pd.Series({
        'k': k_min,
        'grup_berisi_1_orang': grup_1_orang,
        'total_grup': total_grup
    })
```

```python
# TODO F2 – ukur risiko SEBELUM penanganan
kolom_f2 = ['kelurahan', 'jenis_kelamin', 'pendidikan_terakhir', 'tanggal_lahir']
print(cek_k(df, kolom_f2))
```

**Jawaban Anda:** keempat kolom itu tidak memuat nama maupun nomor telepon. Jelaskan dalam dua kalimat mengapa hasilnya tetap berbahaya.

Kombinasi data demografi yang sangat spesifik seperti tanggal lahir lengkap dan kelurahan tempat tinggal memiliki keunikan yang sangat tinggi bagi setiap individu. Meskipun nama dan nomor telepon tidak dicantumkan, penggabungan keempat informasi ini dapat dengan mudah dipasangkan dengan data publik lain (seperti DPT Pemilu atau media sosial) untuk mengidentifikasi pemilik data secara pasti (re-identification attack).

```python
# TODO F3 – ukur risiko SESUDAH generalisasi
kolom_f3 = ['kecamatan', 'jenis_kelamin', 'pendidikan_terakhir', 'kelompok_umur']
print(cek_k(aman, kolom_f3))
```

```python
# TODO F4 – TANTANGAN: capai k >= 5
aman_f4 = aman.copy()

# Strategi: Mengeluarkan kolom 'pendidikan_terakhir' dari variabel quasi-identifier
kolom_f4 = ['kecamatan', 'jenis_kelamin', 'kelompok_umur']

hasil_f4 = cek_k(aman_f4, kolom_f4)
print(hasil_f4)
```

**Jawaban Anda:** tulis kombinasi yang Anda pilih, nilai k yang dicapai, dan informasi apa yang Anda korbankan untuk mencapainya.

- Kombinasi yang dipilih: Menghapus atau melepaskan kolom pendidikan_terakhir dari daftar kombinasi quasi-identifier dan hanya menyisakan ['kecamatan', 'jenis_kelamin', 'kelompok_umur'].   
- Nilai $k$ yang dicapai: $k \ge 5$ (seluruh grup memiliki minimal 5 individu dengan karakteristik yang sama).
- Informasi yang dikorbankan: Granularitas tingkat pendidikan peserta tidak lagi dikombinasikan secara langsung dengan demografi wilayah dan umur saat melakukan analisis privasi, sehingga analisis silang (cross-analysis) yang melibatkan variabel pendidikan dan lokasi menjadi lebih terbatas.

```python
# TODO F5 – sel kecil pada tabel agregat
import pandas as pd

# a. Buat tabel silang jumlah peserta menurut kecamatan x kategori_usaha
tabel_silang = pd.crosstab(aman['kecamatan'], aman['kategori_usaha'])

# b. Hitung berapa sel yang terisi tetapi berisi kurang dari 5 peserta (bernilai 1, 2, 3, atau 4)
sel_kecil = ((tabel_silang > 0) & (tabel_silang < 5)).sum().sum()

# Tampilkan tabel silang dan jumlah sel kecil
print("--- Tabel Silang Kecamatan x Kategori Usaha ---")
print(tabel_silang)

print("\n--- Analisis Risiko Sel Kecil ---")
print(f"Jumlah sel yang berisi kurang dari 5 peserta (1-4): {sel_kecil}")
```
