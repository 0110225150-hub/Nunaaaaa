```python
# TODO E1 — HAPUS
# Buat DataFrame baru bernama 'aman' yang sudah membuang semua kolom bertindakan 'hapus'.
# Cetak berapa kolom yang dibuang dan berapa yang tersisa.

kolom_hapus = tanda.loc[tanda['tindakan'] == 'hapus', 'nama_kolom']

aman = df.drop(columns=kolom_hapus)

print("Jumlah kolom yang dibuang :", len(kolom_hapus))
print("Jumlah kolom yang tersisa :", aman.shape[1])
```

```python
tanda.loc[tanda['nama_kolom'] == 'id_peserta', 'tindakan'] = 'hash'
```

```python
# TODO E2 — HASH
# Tulis fungsi samarkan(nilai, garam) yang mengembalikan 12 karakter pertama
# dari sha256(garam + nilai). Kembalikan np.nan bila nilainya kosong.
# Terapkan pada kolom yang Anda beri tindakan 'hash'.
# Expected: nilai asli dan hasil hash berbeda, dan nilai asli yang sama menghasilkan hash yang sama

try:
    from google.colab import userdata
    GARAM = userdata.get('GARAM')
except Exception:
    GARAM = os.environ['GARAM']

def samarkan(nilai, garam):
    if pd.isna(nilai):
        return np.nan

    teks = str(nilai)
    hasil = hashlib.sha256((garam + teks).encode()).hexdigest()
    return hasil[:12]

# Terapkan pada kolom yang Anda beri tindakan 'hash'.
kolom_hash = tanda.loc[tanda['tindakan'] == 'hash', 'nama_kolom']

for kolom in kolom_hash:
    aman[kolom] = aman[kolom].apply(lambda nilai: samarkan(nilai, GARAM))

print("Kolom yang di-hash:", list(kolom_hash))
```

```python
# TODO E3 — buktikan hash bersifat konsisten dan tidak terbaca
# a. Hash dua kali nilai yang sama, tunjukkan hasilnya identik

nilai1 = 'P001'

hash1 = samarkan(nilai1, GARAM)
hash2 = samarkan(nilai1, GARAM)

print("a. Hash nilai yang sama:")
print("Hash pertama :", hash1)
print("Hash kedua   :", hash2)
print("Hasil identik:", hash1 == hash2)
```

```python
# b. Hash dua nilai berbeda, tunjukkan hasilnya berbeda

nilai2 = 'P002'
hash3 = samarkan(nilai2, GARAM)

print("\nb. Hash nilai yang berbeda:")
print("Hash P001 :", hash1)
print("Hash P002 :", hash3)
print("Hasil berbeda:", hash1 != hash3)
```

_Jika GARAM ikut ter-commit ke GitHub, nilai asli tetap tidak langsung terlihat dari hash. Namun, orang lain yang mengetahui GARAM dapat mencoba menebak nilai asli dan mencocokkannya dengan hash, sehingga keamanan hash menjadi lebih lemah._

```python
# TODO E4 — GENERALISASI

ACUAN = pd.Timestamp('2025-09-01')

aman['usia'] = ACUAN.year - aman['tanggal_lahir'].dt.year

aman.loc[
    (aman['tanggal_lahir'].dt.month > ACUAN.month) |
    (
        (aman['tanggal_lahir'].dt.month == ACUAN.month) &
        (aman['tanggal_lahir'].dt.day > ACUAN.day)
    ),
    'usia'
] -= 1

print(aman['usia'].head())
```

```python
# b. Buat kelompok umur
def kelompok_umur(usia):
    if pd.isna(usia):
        return np.nan
    elif usia < 25:
        return '<25'
    elif usia <= 34:
        return '25-34'
    elif usia <= 44:
        return '35-44'
    elif usia <= 54:
        return '45-54'
    else:
        return '55+'

aman['kelompok_umur'] = aman['usia'].apply(kelompok_umur)

# c. Buang kolom tanggal_lahir dan kelurahan
aman = aman.drop(columns=['tanggal_lahir', 'kelurahan'], errors='ignore')

# d. Rapikan penulisan kecamatan
aman['kecamatan'] = aman['kecamatan'].str.strip().str.title()

# Tampilkan hasil pengecekan
print("Kelompok umur terbanyak :", aman['kelompok_umur'].value_counts().idxmax())
print("Jumlah peserta          :", aman['kelompok_umur'].value_counts().max())
print("Jumlah nilai unik kecamatan :", aman['kecamatan'].nunique())
```

```python
# TODO F1 — tulis fungsi cek_k(data, kolom)
# Kembalikan Series berisi: k (grup terkecil), grup berisi 1 orang, total grup

def cek_k(data, kolom):
    ukuran_grup = data.groupby(kolom).size()

    k = ukuran_grup.min()
    grup_1_orang = (ukuran_grup == 1).sum()
    total_grup = ukuran_grup.size

    return pd.Series({
        'k (grup terkecil)': k,
        'grup berisi 1 orang': grup_1_orang,
        'total grup': total_grup
    })

print(cek_k(aman, 'kecamatan'))
```

```python
# TODO F2 — ukur risiko SEBELUM penanganan
# Gunakan data asli df dengan kolom:
#   ['kelurahan', 'jenis_kelamin', 'pendidikan_terakhir', 'tanggal_lahir']
# Expected: k = 1 dan SELURUH 620 baris berada di grup berisi satu orang
kolom_pengenal = [
    'kelurahan',
    'jenis_kelamin',
    'pendidikan_terakhir',
    'tanggal_lahir'
]

hasil_f2 = cek_k(df, kolom_pengenal)

print("k (grup terkecil)       :", hasil_f2['k (grup terkecil)'])
print("Grup berisi 1 orang     :", hasil_f2['grup berisi 1 orang'])
print("Total grup              :", hasil_f2['total grup'])
print("Jumlah baris            :", len(df))
```

**Jawaban Anda:** keempat kolom itu tidak memuat nama maupun nomor telepon. Jelaskan dalam dua kalimat mengapa hasilnya tetap berbahaya.

_Walaupun tidak ada nama atau nomor telepon, keempat data tersebut jika digabungkan masih bisa menunjukkan ciri-ciri orang tertentu. Data tersebut juga bisa dicocokkan dengan informasi lain untuk mengetahui identitas seseorang._

```python
# TODO F3 — ukur risiko SESUDAH generalisasi
# Gunakan 'aman' dengan kolom:
#   ['kecamatan', 'jenis_kelamin', 'pendidikan_terakhir', 'kelompok_umur']
# Expected: jumlah grup berisi satu orang turun banyak, tetapi k masih 1
kolom_aman = [
    'kecamatan',
    'jenis_kelamin',
    'pendidikan_terakhir',
    'kelompok_umur'
]

ukuran_grup = aman.groupby(
    kolom_aman,
    dropna=False,
    observed=True
).size()

k = ukuran_grup.min()
grup_1_orang = (ukuran_grup == 1).sum()
total_grup = ukuran_grup.size

print("k (grup terkecil)    :", k)
print("Grup berisi 1 orang  :", grup_1_orang)
print("Total grup            :", total_grup)
```

```python
# TODO F4 — TANTANGAN: capai k >= 5

def hitung_k(data, kolom):
    ukuran_grup = data.groupby(
        kolom,
        dropna=False,
        observed=True
    ).size()

    return ukuran_grup.min(), (ukuran_grup == 1).sum(), ukuran_grup.size


# Kondisi awal
kolom_awal = [
    'kecamatan',
    'jenis_kelamin',
    'pendidikan_terakhir',
    'kelompok_umur'
]

print("Kondisi awal:")
print(hitung_k(aman, kolom_awal))


# Pilihan 1 — tanpa jenis_kelamin
kolom_1 = [
    'kecamatan',
    'pendidikan_terakhir',
    'kelompok_umur'
]

print("\nPilihan 1 — tanpa jenis_kelamin:")
print(hitung_k(aman, kolom_1))


# Pilihan 2 — tanpa pendidikan_terakhir
kolom_2 = [
    'kecamatan',
    'jenis_kelamin',
    'kelompok_umur'
]

print("\nPilihan 2 — tanpa pendidikan_terakhir:")
print(hitung_k(aman, kolom_2))


# Pilihan 3 — tanpa kelompok_umur
kolom_3 = [
    'kecamatan',
    'jenis_kelamin',
    'pendidikan_terakhir'
]

print("\nPilihan 3 — tanpa kelompok_umur:")
print(hitung_k(aman, kolom_3))


# Pilihan 4 — umur digeneralisasi menjadi 2 kelompok
aman_f4 = aman.copy()

aman_f4['kelompok_umur_2'] = pd.cut(
    aman_f4['usia'],
    bins=[-float('inf'), 44, float('inf')],
    labels=['<45', '45+']
)

kolom_4 = [
    'kecamatan',
    'kelompok_umur_2'
]

print("\nPilihan 4 — kecamatan + kelompok umur 2 kategori:")
print(hitung_k(aman_f4, kolom_4))


# Pilihan 5 — gabungkan kecamatan kecil
aman_f4['kecamatan_f4'] = aman_f4['kecamatan'].replace({
    'Cipayung': 'Lainnya',
    'Sawangan': 'Lainnya'
})

kolom_5 = [
    'kecamatan_f4',
    'kelompok_umur_2'
]

print("\nPilihan 5 — gabungkan Cipayung dan Sawangan:")
print(hitung_k(aman_f4, kolom_5))
```

**Jawaban Anda:** tulis kombinasi yang Anda pilih, nilai k yang dicapai, dan informasi apa yang Anda korbankan untuk mencapainya.

_Saya memilih kecamatan + kelompok umur (<45 dan 45+), dengan menggabungkan Cipayung dan Sawangan menjadi “Lainnya”. Nilai k yang dicapai adalah 8, dengan mengorbankan sebagian detail wilayah dan umur serta tidak menggunakan jenis kelamin dan pendidikan terakhir._

```python
# TODO F5 — sel kecil pada tabel agregat
# a. Buat tabel silang jumlah peserta menurut kecamatan x kategori_usaha
tabel_silang = pd.crosstab(
    aman['kecamatan'],
    aman['kategori_usaha']
)

print("Tabel silang kecamatan x kategori_usaha:")
print(tabel_silang)
```

```python

# b. Hitung berapa sel yang terisi tetapi berisi kurang dari 5 peserta
# Expected: ada sel bernilai 1-4; sel seperti ini tidak boleh ditampilkan apa adanya di laporan
sel_kecil = tabel_silang[(tabel_silang > 0) & (tabel_silang < 5)]

jumlah_sel_kecil = (sel_kecil > 0).sum().sum()

print("\nJumlah sel terisi dengan kurang dari 5 peserta:", jumlah_sel_kecil)
```

**Penandaan data pribadi:** tidak ditemukan pengenal langsung (nama, NIK, no HP, email, alamat). Risiko ada pada *quasi-identifier* (`Age`, `Gender`, `University_Year`, `Major`) dan gaji. Pada kombinasi `Age + Gender + University_Year + Major`, nilai k = 1 (98 grup berisi satu orang), sehingga `Age` digeneralisasi dan `Student_ID` di-hash pada G4.

```python
# G4 — jalankan penanganan pada data tim dan periksa nilai k
# Salin pola Bagian E dan F, ganti dataset dan nama kolomnya.
# Simpan hasilnya ke data/processed/ dan JANGAN unggah data/raw/

import hashlib
import os
import numpy as np
import pandas as pd

FILE = 'student_career_success_dataset.csv'

if not os.path.exists(FILE):
    try:
        from google.colab import files
        print(f"Silakan unggah berkas '{FILE}' terlebih dahulu:")
        files.upload()
    except ImportError:
        raise FileNotFoundError(
            f"File '{FILE}' tidak ditemukan. Letakkan '{FILE}' di folder yang sama dengan notebook ini."
        )

# 1. Muat data
data_tim = pd.read_csv(FILE)

# 2. Tabel penanganan (sama dengan kamus data G3)
PENANGANAN = {
    'Student_ID': ('pengenal tidak langsung', 'hash'),
    'Age': ('pengenal tidak langsung', 'generalisasi'),
    'Gender': ('data pribadi', 'pertahankan'),
    'University_Year': ('pengenal tidak langsung', 'pertahankan'),
    'Major': ('pengenal tidak langsung', 'pertahankan'),
    'CGPA': ('data pribadi', 'pertahankan'),
    'Starting_Salary_USD': ('data pribadi', 'generalisasi'),
}
for kolom in data_tim.columns:
    PENANGANAN.setdefault(kolom, ('data non-pribadi', 'pertahankan'))

tanda_tim = pd.DataFrame(
    [(k, v[0], v[1]) for k, v in PENANGANAN.items()],
    columns=['nama_kolom', 'kategori', 'tindakan']
)
assert set(tanda_tim['nama_kolom']) == set(data_tim.columns), 'Ada kolom yang belum ada di tabel penanganan'

# 3. Hapus kolom bertindakan 'hapus' (pada dataset ini tidak ada)
kolom_hapus = tanda_tim.loc[tanda_tim['tindakan'] == 'hapus', 'nama_kolom']
data_processed = data_tim.drop(columns=kolom_hapus, errors='ignore').copy()

# 4. Hash: garam diambil dari luar notebook, JANGAN ditulis langsung
try:
    from google.colab import userdata
    GARAM = userdata.get('GARAM')
except Exception:
    GARAM = os.environ['GARAM']

def samarkan(nilai, garam):
    if pd.isna(nilai):
        return nilai
    return hashlib.sha256((garam + str(nilai)).encode()).hexdigest()[:12]

for kolom in tanda_tim.loc[tanda_tim['tindakan'] == 'hash', 'nama_kolom']:
    data_processed[kolom] = data_processed[kolom].apply(lambda x: samarkan(x, GARAM))

# 5. Generalisasi
data_processed['kelompok_umur'] = pd.cut(
    data_processed['Age'], bins=[-np.inf, 21, np.inf], labels=['<=21', '22+']
)
data_processed['kelompok_gaji'] = pd.cut(
    data_processed['Starting_Salary_USD'],
    bins=[-1, 0, 60000, 90000, np.inf],
    labels=['Tidak ada', '<=60rb', '60-90rb', '>90rb']
)
data_processed = data_processed.drop(columns=['Age', 'Starting_Salary_USD'])

# 6. Periksa temuan kualitas yang akan dibawa ke tahap cleaning
print('Jumlah baris setelah penanganan:', len(data_processed))
print('Gaji = 0 (belum bekerja, bukan salah input):', int((data_tim['Starting_Salary_USD'] == 0).sum()))
print('Resume_Score = 100 (kemungkinan batas atas):', int((data_tim['Resume_Score'] == 100).sum()))
print('Senior berusia <= 20 tahun:', int(((data_tim['University_Year'] == 'Senior') & (data_tim['Age'] <= 20)).sum()))

# 7. Periksa k-anonymity
def hitung_k(data, kolom):
    ukuran = data.groupby(kolom, dropna=False, observed=True).size()
    return int(ukuran.min()), int((ukuran == 1).sum()), int(ukuran.size)

print('k kondisi awal:', hitung_k(data_tim, ['Age', 'Gender', 'University_Year', 'Major']))
quasi_identifier = ['kelompok_umur', 'University_Year', 'Major']
nilai_k, grup_1, total_grup = hitung_k(data_processed, quasi_identifier)
print('Quasi-identifier:', quasi_identifier)
print(f'Nilai k: {nilai_k} | grup berisi 1 orang: {grup_1} | total grup: {total_grup}')

# 8. Simpan versi terproses saja
os.makedirs('data/processed', exist_ok=True)
output_file = 'data/processed/student_career_processed.csv'
data_processed.to_csv(output_file, index=False)
print('File hasil penanganan:', output_file)
```

**Catatan nilai k (G4):** setelah penanganan, nilai k = 205 pada quasi-identifier `kelompok_umur`, `University_Year`, dan `Major`. Tidak ada grup yang berisi satu orang (total 64 grup). Sebelum penanganan, k = 1 dengan 98 grup berisi satu orang.
